Pull a Model

```python
#model = "phi:latest"
#model = "mixtral:8x7b-instruct-v0.1-fp16"
model = "gemma2:9b"
model_pull = f"{url_api}pull"

resp = requests.post(model_pull, json=dict(name=model), stream=True)

status_last = None
for line in resp.iter_lines():
    if line:
        status_message = json.loads(line)
        status_current = status_message["status"]
        if status_current != status_last:
            print()
        status_last = status_current
        print(f"\r{status_current}", end="")
```
